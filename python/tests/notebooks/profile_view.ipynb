# A query profile in a notebook

A `ProfileView` left as the last expression of a cell draws itself: the plan as a tree, each operator's time and
rows, and the pipelines on a timeline. `tests/test_notebooks.py` runs this notebook with nbval, which compares each
cell's output with the one stored here.

The first profile is a stored document, TPC-H Q9 at scale factor 0.01, so its picture is the same on every run.

```python
import json
from pathlib import Path

from krishiv.sql.profile_view import ProfileView

view = ProfileView(json.loads(Path("../fixtures/q09-profile.json").read_text()))
view
```

`krishiv.ui(df)` runs the query and returns the same view of its profile. The times differ from run to run, so
this cell's output is not compared; the cell after it says what every run must show.

```python
# NBVAL_IGNORE_OUTPUT
import krishiv
from krishiv.sql import SparkSession

spark = SparkSession.builder.create()
df = spark.sql("SELECT id % 3 AS k, count(*) AS n FROM range(1000) GROUP BY id % 3")
live = krishiv.ui(df)
live
```

```python
html = live._repr_html_()
print(type(live).__name__, "Aggregate" in html, "<script" in html)
spark.stop()
```
